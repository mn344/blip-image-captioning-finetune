```python
import os, json, random

IMAGE_DIR = "/content/all_images"
CAPTIONS_FILE = "/content/captions_final.txt"
TRAIN_JSON = "/content/train.json"
VAL_JSON = "/content/val.json"

entries = []
with open(CAPTIONS_FILE, "r", encoding="utf-8") as f:
    for line in f:
        line = line.rstrip("\n")
        if "\t" not in line:
            continue
        fname, caption = line.split("\t", 1)
        fname, caption = fname.strip(), caption.strip()
        if os.path.exists(os.path.join(IMAGE_DIR, fname)) and caption:
            entries.append({"file_name": fname, "caption": caption})

print(f"Total valid entries: {len(entries)}")

random.seed(42)
random.shuffle(entries)
val_size = max(1, int(len(entries) * 0.10))
val_entries = entries[:val_size]
train_entries = entries[val_size:]

json.dump(train_entries, open(TRAIN_JSON, "w"), indent=2, ensure_ascii=False)
json.dump(val_entries, open(VAL_JSON, "w"), indent=2, ensure_ascii=False)
print(f"Train: {len(train_entries)} | Val: {len(val_entries)}")
```

```python
import torch
from PIL import Image
from torch.utils.data import Dataset
from transformers import BlipProcessor, BlipForConditionalGeneration, TrainingArguments, Trainer

MODEL_NAME = "Salesforce/blip-image-captioning-base"
OUTPUT_DIR = "/content/drive/MyDrive/blip_finetuned_v3_checkpoints"

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

processor = BlipProcessor.from_pretrained(MODEL_NAME)
model = BlipForConditionalGeneration.from_pretrained(MODEL_NAME).to(device)

class CaptionDataset(Dataset):
    def __init__(self, json_path, image_dir, processor):
        self.entries = json.load(open(json_path))
        self.image_dir = image_dir
        self.processor = processor

    def __len__(self):
        return len(self.entries)

    def __getitem__(self, idx):
        item = self.entries[idx]
        image = Image.open(os.path.join(self.image_dir, item["file_name"])).convert("RGB")
        encoding = self.processor(images=image, text=item["caption"], padding="max_length",
                                   truncation=True, max_length=64, return_tensors="pt")
        encoding = {k: v.squeeze(0) for k, v in encoding.items()}
        labels = encoding["input_ids"].clone()
        labels[labels == self.processor.tokenizer.pad_token_id] = -100
        encoding["labels"] = labels
        return encoding

train_dataset = CaptionDataset(TRAIN_JSON, IMAGE_DIR, processor)
val_dataset = CaptionDataset(VAL_JSON, IMAGE_DIR, processor)
print(f"Train examples: {len(train_dataset)} | Val examples: {len(val_dataset)}")

training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=10,
    learning_rate=3e-5,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=2,
    logging_steps=20,
    fp16=torch.cuda.is_available(),
    load_best_model_at_end=True,
    metric_for_best_model="eval_loss",
    greater_is_better=False,
    weight_decay=0.01,
    report_to=[],
)

trainer = Trainer(model=model, args=training_args, train_dataset=train_dataset, eval_dataset=val_dataset)
trainer.train()
```

```python
final_dir = os.path.join(OUTPUT_DIR, "final_model")
model.save_pretrained(final_dir)
processor.save_pretrained(final_dir)
print(f"Saved to: {final_dir}")

import shutil
shutil.make_archive("/content/drive/MyDrive/blip_finetuned_v3_final", 'zip', final_dir)
print("Zip saved to Drive!")
```
